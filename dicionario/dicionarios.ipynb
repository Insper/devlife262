# Dicionarios em Python

Um dicionario guarda pares do tipo **chave: valor**. Ele e muito util quando queremos acessar uma informacao pelo nome dela, e nao pela posicao.

## Criando um dicionario

```python
aluno = {
    "nome": "Ana",
    "idade": 18,
    "curso": "Engenharia"
}

aluno
```

## Acessando valores

Para pegar um valor, usamos a chave entre colchetes.

```python
print(aluno["nome"])
print(aluno["curso"])
```

Tambem podemos usar `.get()`. Ele evita erro quando a chave nao existe.

```python
print(aluno.get("idade"))
print(aluno.get("telefone", "telefone nao cadastrado"))
```

## Alterando e adicionando valores

```python
aluno["idade"] = 19
aluno["email"] = "ana@email.com"

aluno
```

## Removendo valores

```python
email_removido = aluno.pop("email")

print(email_removido)
print(aluno)
```

## Verificando se uma chave existe

```python
if "nome" in aluno:
    print("O aluno tem nome cadastrado")

if "telefone" not in aluno:
    print("O aluno ainda nao tem telefone cadastrado")
```

## Percorrendo um dicionario com `for`

```python
for chave in aluno:
    print(chave)
```

```python
for valor in aluno.values():
    print(valor)
```

```python
for chave, valor in aluno.items():
    print(chave, "->", valor)
```

## Exemplo: notas de uma turma

```python
notas = {
    "Ana": 8.5,
    "Bruno": 6.0,
    "Carla": 9.2,
    "Diego": 7.5
}

soma = 0

for nota in notas.values():
    soma += nota

media = soma / len(notas)
print("Media da turma:", media)
```

## Exemplo: contando palavras

```python
frase = "python e legal e python e poderoso"
contagem = {}

for palavra in frase.split():
    if palavra not in contagem:
        contagem[palavra] = 0
    contagem[palavra] += 1

contagem
```

## Dicionarios com listas e outros dicionarios

```python
produto = {
    "nome": "Caderno",
    "preco": 24.90,
    "estoque": 12,
    "categorias": ["papelaria", "escola"],
    "dimensoes": {
        "altura": 28,
        "largura": 20
    }
}

print(produto["categorias"][0])
print(produto["dimensoes"]["altura"])
```

## Metodos comuns

| Metodo | O que faz |
| --- | --- |
| `d.keys()` | Retorna as chaves |
| `d.values()` | Retorna os valores |
| `d.items()` | Retorna pares `(chave, valor)` |
| `d.get(chave, padrao)` | Busca uma chave sem gerar erro |
| `d.pop(chave)` | Remove uma chave e retorna seu valor |
| `d.update(outro_dicionario)` | Atualiza ou adiciona varios pares |

```python
aluno.update({
    "curso": "Ciencia da Computacao",
    "semestre": 1
})

aluno
```
