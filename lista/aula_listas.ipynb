| Método                              | Descrição                                                                                                                                       |
| ----------------------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------- |
| `append(x)`                         | Adiciona o elemento `x` ao final da lista.                                                                                                      |
| `extend(iterable)`                  | Adiciona ao final da lista todos os elementos de um iterável, como outra lista, tupla ou `range`.                                               |
| `insert(i, x)`                      | Insere o elemento `x` na posição de índice `i`.                                                                                                 |
| `remove(x)`                         | Remove a **primeira ocorrência** do elemento `x`. Gera `ValueError` se o elemento não estiver na lista.                                         |
| `pop(i=-1)`                         | Remove e retorna o elemento da posição `i`. Se nenhum índice for informado, remove o último elemento.                                           |
| `clear()`                           | Remove todos os elementos da lista, deixando-a vazia.                                                                                           |
| `index(x, start=0, end=len(lista))` | Retorna o índice da **primeira ocorrência** de `x`. Pode limitar a busca usando `start` e `end`. Gera `ValueError` se não encontrar o elemento. |
| `count(x)`                          | Retorna quantas vezes o elemento `x` aparece na lista.                                                                                          |
| `sort(*, key=None, reverse=False)`  | Ordena os elementos da própria lista. Modifica a lista original e retorna `None`.                                                               |
| `reverse()`                         | Inverte a ordem dos elementos da própria lista. Modifica a lista original e retorna `None`.                                                     |
| `copy()`                            | Retorna uma cópia superficial (*shallow copy*) da lista.                                                                                        |
